# Logical Operators

and/or/not operators

```python
# conditions with and/or
credits = 44
gradeLevel = 10
if credits >=44 and gradeLevel >=11:
  print("graduate 1st if statement")

if credits >=44 or gradeLevel >=11:
  print("graduate 2st if statement")
```

```python
# lets we want to check for if a has a value in the range between 40 and 60 inclusive
a = int(input("Enter an integer: "))
if a>= 40 and a<=60:
  print(f'the number {a} is between 40 and 60 inclusive')

# We can also write the boolean expression of range using tradition math notation
if 40<=a<=60:
  print(f'the number {a} is between 40 and 60 inclusive')

#what if we want to check if it's not in the range
if not(a>= 40 and a<=60):
  print(f'the number {a} is NOT between 40 and 60 inclusive')

if not(40<=a<=60):
  print(f'the number {a} is NOT between 40 and 60 inclusive')
```
