Who can we write a program that can determine if an integer is a prime number?


```python
 #Method 1
 # 22801763489

divisor = 2
num = int(input("Enter an integer"))

while not(num % divisor == 0):
  divisor+=1

if num == divisor:
  print(f'{num} is a prime number')
else:
  print(f'{num} is not a prime number because it divisible by {divisor}')
```
