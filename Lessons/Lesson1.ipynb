```python
term = int(input("How many terms for the Leibniz series? "))
sum = 0
sign = 1
den = 1

for x in range(term):
    term = sign * (1 / den)
    sum += term
    sign *= -1
    den += 2

ans = sum * 4
print(ans)
```

```python
term=int(input("which fibonacci term?"))
fib1=0
fib2=1
if term==1:
    print(fib1)
elif term==2:
    print(fib2)
elif term>=3:
    for x in range(term-2):
        newfib = fib1 + fib2
        fib1 = fib2
        fib2 = newfib
    print(newfib)
```
