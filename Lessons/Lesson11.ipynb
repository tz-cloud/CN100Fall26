# **Functions**

A Function is a useful way dividing up and organizing your code into useful blocks.

Allows you reuse your code, save time, and makes your code more readable

is a keyway to defining interfaces so code can be shared

functions can be passed information that it needs to do it's task in a form of parameters (Optional). Parameters are variables that take the information passed to it. Parameters (if more than on) are seperated by commas.

Generally functions return information, but some functions don't return anything (optional)

If function does return information, it done with return statement. If a function doesn't return anything, then return statement is optional.

Varaibles created in functions only exist in the function and once the function returns, the variables are destroyed. (Scope of a variables)

SYNTAX:

def function_name(parameter,...):

(indent) code_body

...

code_body return **return_info**

```python
#welcome
def welcome():
  print("Hello...")

#personlized greeting
def welcome2(name):
  print(f"Welcome {name} to Python...")

#Area of a triangle
def triangle_area(b,h):
  area = 0.5*b*h
  return area

#roll two dice
import random as rnd
def roll_dice():
  dice1 = rnd.randint(1,6)
  dice2 = rnd.randint(1,6)
  return dice1,dice2
```

```python
d1,d2 = roll_dice()
print(d1,d2)
```

```python
def isPrime(num):
  divisor = 2
  sqrt_num = round(num**(0.5))
  while not(num % divisor == 0 or sqrt_num < divisor):
    divisor+=1

  #return sqrt_num < divisor

  if sqrt_num < divisor:
    return True
  else:
    return False
```

```python
#print( isPrime(105) )

if( isPrime(15) ):
  print("is prime")
else:
  print("not prime")
```
