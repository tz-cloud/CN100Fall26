## Range function

range(start,end,step)


*   start - default 0 if not provided
*   end - must be provided
*   step - default is 1 if not provided

range(n) - generates a Seqeunce of numbers from [0,n)

```python
for x in range(10):
  print("hello")
```

```python
for x in range(1,10):
  print(x)
```

```python
for x in range(-5,10):
  print(x)
```

```python
for i in range(0,8,2):
  print(i)
```

```python
for x in range(9,0,-1):
  print(x)
```

```python
#sum up the number from 1 - 10
sum = 0
for x in range(1,11):
  sum+=x

print(f'Sum is {sum}')
```

```python
#sum up the number from 1 - 101 that are odd numbers
sum = 0
for x in range(1,102,2):
  sum+=x

print(f'Sum is {sum}')
```

```python
#sum of the squares from 1 - 10
sum = 0
for x in range(1,11):
  sum+=x**2

print(f'Sum is {sum}')
```

```python
#sum of the square roots from 1 - 10
sum = 0
for x in range(1,11):
  sum+=x**(1/2)

print(f'Sum is {sum}')
```
