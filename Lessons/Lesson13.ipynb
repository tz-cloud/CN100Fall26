# **len() function**
is a built in function returns the number of items in a structure.

- For strings, it return the number of characters in a string

```python
#empty string
s = ''
print( len(s) )

s="i am"
print( len(s) )
```

find() string function
the function returns the index position of the first instatance, from left to right, of a character or string of characters,

otherwise, -1 is returned if not found

```python
msg = 'I am who I am'
print( msg.find('I') )
print( msg.find('am') )
print( msg.find('i') )
```

# **Transversing a string**

-Checking how many occurrance of "the" in a string

```python
msg = 'I am who I am'
for i in range(len(msg)):
  print(msg[i])
```

```python
s = "The cat in the hat"
for char in s:
  print(char)
```

```python
# How can we print out 3 characters at a time?
msg = "the cat in the hat ate my hat"
for i in range(0, len(msg), 3):
  letters = msg[i:i+3]
  print(letters)
```

```python
msg = "the cat in the hat ate my hat"
count=0
for i in range(0, len(msg)):
  letters = msg[i:i+3]
  if letters == 'the':
    count+=1

print(count)
```

```python
#Python function that counts number of occurances
txt = "I love apple, apple are my favorite fruit"
num = txt.count("apple")
print(num)
```

```python
#replace all the apple with banana
txt = "I love apple, apple are my favorite fruit"
new_txt = txt.replace("apple","banana")
print(new_txt)
print(txt)
txt = new_txt
print(txt)
```
