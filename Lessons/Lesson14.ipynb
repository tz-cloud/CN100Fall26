How are characters represent in the digital worlds?

Since digital devices are circuits, ie- on/off switches which can be represented as 0-OFF and 1-ON, we call this a bit. We can group these bits into groups of 8 bits called a byte.

These bytes mathematically represent binary number for example 00001001 = 9 as a decimal number. So all characters we see on our screens are digitally represented as an integer value such. These values are called ASCII Values:

[ASCII Table](https://www.asciitable.com/)

Standard ASCII values are from 0-255

```python
# function called ord() convert an character to it's ascii numberical representation
ascii = ord('A')
print( ascii)

ch = chr(97)
print(ch)


msg = 'ATTACK'
for ch in msg:
  ascii = ord(ch)
  print(f' {ch}-->{ascii}')
```

```python
#How do we take an ascii numerical number and convert it to a character
print( chr(65))
```

```python
def encode(msg):
  bld=''
  for ch in msg:
    ascii = ord(ch)
    new_ch = chr(ascii+1)
    bld += new_ch

  return bld
```

```python
def decode(msg):
  bld=''
  for ch in msg:
    ascii = ord(ch)
    new_ch = chr(ascii-1)
    bld += new_ch

  return bld
```

```python
enc = encode("ATTACK-Z")
print(enc)
dec = decode(enc)
print(dec)
```
