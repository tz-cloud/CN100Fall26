While Loops

Repeat a block of while the condition is true

while {condition}:
  code
  code


So we use a while when we are not sure the number of times we to repeat a block of code.

```python
x=0
while x<7:
  x+=1
  print("hello")
```

In the example we want to keep saying hello until the user enters 'stop'

```python
#Demo of keep looping till the condition is no longer true
response = 'x'
while response != 'stop':
  print("Continue with game")
  response = input("Enter stop to end game...")
```

```python
import random as rnd
print(rnd.randint(1,10))
```

```python
#guessing game from 1 to 10
secret_num = rnd.randint(1,10)
print(secret_num)
num_of_tries = 1
guess = int(input("Guess a number from 1 to 10 "))
while not(guess == secret_num or num_of_tries >= 3):
  guess = int(input("Incorrect, try again... guess a number from 1 to 10 "))
  num_of_tries+=1

if guess == secret_num:
  print(f'You guessed correctly!!')
else:
  print("Sorry...you have run out of tries...")
```
