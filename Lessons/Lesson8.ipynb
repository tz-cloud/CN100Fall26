# Applications of using for loops

Fibonacci sequence

[Fibonacci sequence link](https://www.mathsisfun.com/numbers/fibonacci-sequence.html)

```python
nth_num = int(input("Which fibonacci number do you want"))
first =1
second =1
if nth_num == 1:
  print(f'the {nth_num}th fibonacci number is 1')
elif nth_num == 2:
  print(f'the {nth_num}th fibonacci number is 1')
else:
  for x in range(nth_num-2):
    fib = first+second
    first = second
    second = fib

  print(f'the {nth_num}th fibonacci number is {fib}')
```

Approximating PI  3.14159265359

```python

denom = 1
sum=0
for x in range(10000000):
  if x%2 ==0:
    sum += 1/denom
  else:
    sum -= 1/denom
  denom+=2

print(4*sum)
```

How can we estimate sin(x) using a infinite series

image.png

```python
import math

N= int(input("how many terms to estimate"))
x=math.pi/4
sum =0

for k in range(N+1):
  term= (-1)**k/math.factorial(2*k+1)*x**(2*k+1)
  sum+= term
  #print(f'Term-{k+1} is {term}')

print(f'sin({x}) is approximately {sum}')

print(f'Calculator value: {math.sin(math.pi/4)}')


```
