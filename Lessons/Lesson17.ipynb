```python
#How can we store this table in python?

table = [
          [1,100,2.23],
          [3,50,10.99],
          [5,223,54.00]
        ]
#table is a list with each element of a list being another list
#[row][column]
print(table[0])
print(table[1])
print(table[1][1])

```

```python
for row in table:
  print(row)
  print(len(row))
```

```python
#this is method is good if you just access the values in the table
#and not change any values in the table
for row in table:
  for col in row:
    print(col)
```

```python
#display each number in the table start



for row in range(len(table)):
  for col in range(len(table[row])):
    print(f'Row: {row}  Col: {col}')
    print(table[row][col])
```

```python
# We use this method if need change the values
# in the table.

for row in range(len(table)):
  for col in range(len(table[row])):
    table[row][col]*=-1

for row in range(len(table)):
  print(table[row])
```
