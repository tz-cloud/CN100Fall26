**Strings**

In python, strings are surrounded by single or double quotes.

Strings are immutable, which means you can not change the string once it's been assigned to a variable.

String are an array(list) of characters.

Sting concatenation only joins two strings to make a bigger string. It will not convert non string to a strings, for that you have to convert non-string implicitly to a string with str() function

the "in" built in function of python

the "in" function return a true or false (boolean)

You can search if a string contains another string

```python
msg = 'The cat ate my hat'
print( 'cat' in msg)
print( 'the' in msg)
```

"not in" python built in function

Checks if some string is not in another string. The function returns true or false

```python
msg2 = "Everyone loves money"
if 'dollars' in msg2:
  print('yes')
else:
  print('no')


if 'dollars' not in msg2:
  print("Dollars is in msg2")
```

**String Slicing** is how we extract/copy part of an existing string

Slicing is very powerful concept in python which is used in lists and tuples(we'll see later)

```python
alpha = 'abcdef'
print( alpha[1:5] )  #index [start:end] the last character of index 5 is not included
```

```python
#how do we get the last character of the string

#case1 if we know the what the string is
print( alpha[5])

#case2, we don't know the what the string is ***this most common scenerio you will be dealing with
```

```python
print( alpha[:5] ) #default for start index if blank, is zero
print( alpha[3:])  #default for end index if blank, the end of the string
```

```python
print( alpha[1:6:3] )
```

```python
#Strange but useful
alpha = 'abcdef'
print( alpha[-1:])
print( alpha[-2:])
print( alpha[-5:])
print( alpha[-6:])
print( alpha[-10:])
```
