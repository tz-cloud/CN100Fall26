What is a dictionary?

Dictionaries are a way to store data as a key:value pairs.

Dictionaries do not allow duplicates.

```python
list1 = [1,2,"wolf",(123,321,0)]
#access the string "wolf" we use the index 2, for example
print(list1[2])
```

```python
dict1 ={
          'roy':70,
          23:65,
          'wenqi':list1,
          'bayside':"In Queens"
       }
```

```python
#Accessing data in a dictionary
print(dict1['bayside'])
key='bayside'
print(dict1[key])
print(dict1[23])
print(dict1['wenqi'])
print(dict1['wenqi'][2])
print(dict1['roy'])

```

```python
person1 = {
            'name':'Tom',
            'age': 10,
            'address':'Bayside NY'
          }
```

```python
print(person1['name'])
print(person1['age'])
```

```python
#How do we transverse a dictionary so we can access each element in the dictionary
#for key in person1:
  #print(key)

#print("-"*25)

for key in person1:
  #print(key)
  print(f'Key is {key} and value of person1[{key}] : {person1[key]}')
```

```python
# A dictionaries of individuals where the values is another dictionary
students = {
            "Tom": {'age':14,'grade':10,"address":"Flushing NY"},
            "Jerry":{'age':15, 'grade':11, "address":"Bayside NY"},
            "Bug Bunny":{'age':12, 'grade':11, "address":"SmallHole NY"}
          }

```

```python
print(students['Jerry'])
jerry = students['Jerry']
print(jerry['age'])
print(students['Jerry']['age'])
```

```python
#lets get each student
for student in students:
  print(f'{student} is the key')
  s1 = students[student]
  print(f'address is {s1["address"]}')





```

```python
# list of dictionaries
student_list = []
student_list.append({"name":"Tom", 'age':14,'grade':10,"address":"Flushing NY"})
student_list.append({"name":"Jerry",'age':15, 'grade':11, "address":"Bayside NY"})
print(student_list[0])
print(student_list[1])
```

```python
for student in student_list:
  print(student)
```
