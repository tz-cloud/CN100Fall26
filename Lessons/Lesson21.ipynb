# **PIL - Python Image Library**

The Python Imaging Library adds image processing capabilities to your Python interpreter.

This library provides extensive file format support, an efficient internal representation, and fairly powerful image processing capabilities.

The core image library is designed for fast access to data stored in a few basic pixel formats

```python
from PIL import Image

#Open an existing image file and store the image in a variable
pic = Image.open("/content/drive/MyDrive/Porchetta/CollegeNow/projects/ImageProcessing/fractal.png")

pic
```

```python
#to make a copy of a image and save to another variable
copy_of_pic = pic.copy()


#take a store imaged in a variable and save to a file
copy_of_pic.save("temp.png")
```

```python
#to load the matrix structure 2d array where each element is tuple that store the pixel information RBGA model)
pixPic = pic.load()

print(pixPic)
```

```python
# To get the dimension of the matrix

print( pic.size)
xSize,ySize = pic.size
print(xSize,ySize)
```

```python
#first pixle at position [0,0]
pix = pixPic[0,0]
print(pix)
```

```python
#Lets get the center pixel in the matrix
pixCenter = pixPic[xSize/2,ySize/2]
print(pixCenter)
#Lets print out the value of each component of the center pixel
print(f'Value of red in the center pixel is {pixCenter[0]}')
print(f'Value of green in the center pixel is {pixCenter[1]}')
print(f'Value of blue in the center pixel is {pixCenter[2]}')
print(f'Value of alpha in the center pixel is {pixCenter[3]}')
print('-'*20)
# or you can get the r,g,b,a this alternate way
r,g,b,a = pixCenter
print(f'Value of red in the center pixel is {r}')
print(f'Value of green in the center pixel is {g}')
print(f'Value of blue in the center pixel is {b}')
print(f'Value of alpha in the center pixel is {a}')
```

```python
#change the color of blue to 250 from 78
pixPic[xSize/2,ySize/2] = (r,g,250,a)
print(pixPic[xSize/2,ySize/2])
```

```python
# all the pix green values to 255

copy_pic = pic.copy()
pixMatrix = copy_pic.load()
rowSize, colSize = copy_pic.size
for row in range(rowSize):
  for col in range(colSize):
    r,g,b,a = pixMatrix[row,col]
    pixMatrix[row,col] = (r,255,b,a)


copy_pic
```

```python
# all the pix green values to 255 if the red is greater than 150

copy_pic = pic.copy()
pixMatrix = copy_pic.load()
rowSize, colSize = copy_pic.size
for row in range(rowSize):
  for col in range(colSize):
    r,g,b,a = pixMatrix[row,col]
    if r>150:
      pixMatrix[row,col] = (r,255,b,a)


copy_pic
```

```python
# all the pix to a grey color if the red is greater than 150
copy_pic = pic.copy()
pixMatrix = copy_pic.load()
rowSize, colSize = copy_pic.size
for row in range(rowSize):
  for col in range(colSize):
    r,g,b,a = pixMatrix[row,col]
    if r>150:
      pixMatrix[row,col] = (128,128,128,a)


copy_pic
```
