**What is a list? **


A list is data type that allows us to store multple items under a single variable name.

Elements is a general term used to referenece the items in the list.

In Python, a list is used to store a collection of data, which we can add to (increasing the size) or remove items from (decrease the size) a list.

List start and end with square bracks and we use a comma to seperate elements

```python
# How to create an empty list
fruits = []
print(len(fruits))

grades = [90,89,32,76,100]
print( grades)
print( len(grades) )
```

```python
#mixed bag of element (ie. a list of different data types)
mix_bag = [1234, 'jack',3.14, [54,31.2, 'string'] ]
print( len(mix_bag))
#how we access elements of a list
print( mix_bag[0])
print( mix_bag[1])
print( mix_bag[3])
tmp =mix_bag[3]
print(tmp)
print(tmp[2])
print( mix_bag[3][2])
#print( mix_bag[3][2])
```

```python
#How do we go or transverse through each element in the list
grades = [90,89,32,76,100]

#two ways
# use this for lop method if you do NOT want to change
# values in the list grades
for element in grades:
  print(element)

print('-'*10)

#Use this Method if you are planning to change values
#in the list grades.
for idx in range(len(grades)):
  print(grades[idx])
```

```python
# This example demonstrates which method to use
# when want to make changes to the values of a list.
grades1 = [90,89,32,76,100]
grades2 = [90,89,32,76,100]
for element in grades1:
  element+=2
print(grades1)

for idx in range(len(grades2)):
  grades2[idx]+=2
print(grades2)
```

```python
print(f'get last element of the list {grades1[-1]}')
print(f'get last element of the list {grades1[len(grades1)-1]}')
```

How do add and remove elements from a list?

```python
fruits = []
#adding an element at the end of a list
fruits.append("apple")
print(fruits)
fruits.append('orange')
fruits.append('strawberry')
print(fruits)
```

```python
#Second way of adding an element at a certain position in the list
#using the index
fruits.insert(1,'banana')
print(fruits)
```

```python
#removing an elemtn in a list
fruits=['apple','banana','orange','strawberry',"orange"]
fruits.remove("orange")
print(fruits)

#second method is by position using the index
del fruits[1]
print(fruits)
```

```python
# we can delete a range of elements in a list using del and indice like in slicing a string
grades = [90,89,32,76,100]
del grades[1:3]
print(grades)
```

```python
def avg(grds):
  sum = 0
  for grade in grds:
    sum+=grade

  return sum/len(grds)
```

```python
# test the function avg()
grades = [90,89,32,76,100]
grd_avg = avg(grades)
print(grd_avg)
```
