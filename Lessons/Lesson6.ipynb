#Traffic Light Example#

```python
# Flip a coin
coin = input("Enter H-heads or T-tails")
if coin == 'T' or coin == 't':
  print("win")
if coin == 'H' or coin == 'h':
  print("lose")

#So when event/condition can have two possible out comes we can use
# if...else for efficiency
if coin == 'T' or coin == 't':
  print("win")
else:
  print("lose")

```

```python
light = input("Enter traffic light color: R-red, Y-yellow, G-green: ")
if light == "R" or light == 'r':
  print("Red Light")
elif light == "G" or light == 'g':
  print("Green Light")
else:
  print("Yellow Light")
```

#Shipping cost example#

Shipping cost:

0 - 5 (inclusive) lbs : $5

5 - 10 (inclusive) lbs : $8

greater 10 lbs: $8 + 0.1 per pound over 10 lbs.

```python
weight = float(input("Enter shipping weight (lbs): "))
valid_weight = True

if 0 < weight <= 5:
  shipping_cost = 5

elif 5 < weight <= 10:
  shipping_cost = 8

elif weight > 10:
  shipping_cost = 8 + (weight-10)*0.1
else:
  print("Invalid weight")
  valid_weight = False

if valid_weight:
  print(f'Our weight is {weight} lbs. and the shipping is ${shipping_cost}')
```
