Tuples

What is a tuple
It's almost like a list, it store many different datatypes under single name

You can access the elements in a tuple exactly the way you access an element in a list using square brackets and index number

What's the difference between a list and tuple

- you create a tuple with ()
- once a tuple is create, you can not change it's values nor its size (i.e., add or removing elements)
- It superior in performace (ie, speed).

```python
tup1 = (10, "tuple", 2.234,(1,2,3),[3,4,5])

print( tup1[0])
print( tup1[1])
print( tup1[3])

print( tup1[3][1])

#can't change the value of an element in a tuple

#tup1[0]=-1
```

```python
#We can string together a sequence of element of similar
#or different datatypes to create a tuple

x=30
y=2.5
name = 'Dante'

tup2 = x,y,name
print( tup2)

```

```python
#We transverse a tuple similar to a list.  We can use the index or get each element
#(ie, the two different for loops)

#using the index

for idx in range(len(tup2)):
  print(tup2[idx])

print('-'*10)
#get a copy of each element
for element in tup2:
  print(element)
```

```python
#We can convert a list to a tuple and if we need make a change to a value
# value in the tuple we can us the list fruit, change the value to
# 'orange' and then create a new tuple

fruit =['grapes','apples','banana']
tup1 = tuple(fruit)
print(fruit,tup1)

fruit[1] = 'oranges'
tup1 = tuple(fruit)
print(fruit,tup1)

#my perferred method.
tup1 = 'mango',tup1[1],tup1[2]
print(tup1)
```

```python
#Change a tuple within a list, for example replace 2nd tuple
#by leaving the first and last element the same, but
#changing the middle (2nd element) by a multiple of 2

myList = [(1,2,3), (4,5,6),(7,8,9)]

a1 = myList[1][0]
a2 = myList[1][1]*2
a3 = myList[1][2]
print(a1,a2,a3)
myList[1]= a1,a2,a3

print(myList)

#Alternate
myList = [(1,2,3), (4,5,6),(7,8,9)]

myList[1]= myList[1][0], myList[1][1]*2, myList[1][2]
print( myList)
```
